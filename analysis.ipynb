```python
from pathlib import Path
import pandas as pd

# 1. Ruta relativa al archivo desde la raíz del proyecto
json_path = Path("fightmatrix/fightmatrix/spiders/dwcs.json")

# Cargar las stats de los peleadores directamente desde el archivo JSON
df_fighters = pd.read_json(json_path)

# 2. Definir o cargar la lista de matchups
matchups_json = [
    {"matchup": "Patrick Rivera vs Adam Darby"},
    {"matchup": "Brandon Holmes vs Modestino Rodrigues"},
    {"matchup": "Hunter Smith vs Adam Livingston"},
    {"matchup": "Liam McCracken vs Silvestre Sanchez"},
    {"matchup": "Gabriel Lourenco vs Charlie Cleveland"},
]
df_matchups = pd.DataFrame(matchups_json)

# 3. Separar el campo "matchup" en dos columnas de nombres
df_matchups[["fighter_1", "fighter_2"]] = df_matchups["matchup"].str.split(
    " vs ", expand=True
)

# 4. Cruzar la información del Peleador 1 y Peleador 2 lado a lado
df_merged = df_matchups.merge(
    df_fighters.add_suffix("_f1"),
    left_on="fighter_1",
    right_on="fighter_f1",
    how="left",
).merge(
    df_fighters.add_suffix("_f2"),
    left_on="fighter_2",
    right_on="fighter_f2",
    how="left",
)

# Mostrar el resultado final
df_merged
```

```python
def comparar_matchup(df_merged, index=0):
    # 1. Obtener la fila del matchup deseado
    row = df_merged.iloc[index]

    f1_name = row["fighter_1"]
    f2_name = row["fighter_2"]

    # 2. Identificar qué estadísticas existen (quitando los sufijos _f1 / _f2)
    stats = [
        col.replace("_f1", "")
        for col in df_merged.columns
        if col.endswith("_f1") and col != "fighter_f1"
    ]

    # 3. Armar un nuevo DataFrame comparativo
    datos_comparativos = []
    for stat in stats:
        datos_comparativos.append(
            {
                "Estadística": stat,
                f1_name: row[f"{stat}_f1"],
                f2_name: row[f"{stat}_f2"],
            }
        )

    df_comp = pd.DataFrame(datos_comparativos).set_index("Estadística")
    print(f"=== {row['matchup']} ===")
    return df_comp


# Ejemplo: Comparar la primera pelea (índice 0)
comparar_matchup(df_merged, index=3)
```
